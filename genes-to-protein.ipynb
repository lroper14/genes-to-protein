### Get file function

```python
# Function that takes file and removes /t
def get_codon_dict(filename):
    file = open(filename)

    aa_dict = {}

    # Removes new line and tab. 'line' is a list of the codon and AA names. Add to dictionary aa_dict
    for line in file.readlines():
        line = line.strip()
        line = line.split('\t')
        aa_dict[line[0]] = {'1-letter': line[1], '3-letter': line[2], 'name': line[3]}
    
    file.close()
    return aa_dict
```

```python
codon_dict = get_codon_dict('genetic_code.tsv')
print(codon_dict)
```

### Dictionary of exon start and end points within lactase gene

```python
def get_exon_index_dict(filename):
    
    file = open(filename)
    
    exon_index_dict = {}
    results = []
    
    # Remove \n and make list of exon indexes
    for line in file.readlines():
        line = line.strip()
        line = line.split('\t')

        # Add each exon number to a start and end key
        exon_index_dict['start'] = int(line[0])
        exon_index_dict['end'] = int(line[1])

        # Add this dictionary to the list
        results.append(exon_index_dict)

        # Reset the dictionary
        exon_index_dict = {}
    
    file.close()
    return results
```

```python
exon_indexes = get_exon_index_dict('lactase_exon.tsv')
```

### Obtain DNA sequence from file

```python
def get_sequence(filename):
    
    file = open(filename)

    sequence = ''
    for line in file.readlines():
        line = line.strip()
        sequence += line
    
    return sequence
```

```python
sequence = get_sequence('lactase_gene.txt')
```

### Construct the mRNA sequence from the DNA sequence

```python
def construct_mrna(dna = sequence, exon_dict = exon_indexes):
    
    exon_sequence = ''
    
    for dicts in exon_dict:
        exon_sequence += dna[dicts['start']:dicts['end']]

    mrna = exon_sequence.replace('T', 'U')
    return mrna
```

```python
mrna = construct_mrna(sequence, exon_indexes)
```

### Convert the mRNA to protein

```python
def convert_mrna_to_protein(mrna, codon_dict = codon_dict):
    # Finds the position of the start codon
    start_position = mrna.find('AUG')

    # Counter to make sure the while loop ends
    counter = 0 + start_position
    protein = ''

    while counter < len(mrna):
        # Substring of mRNA by finding the position plus 3 for a three
        # letter codon
        codon = mrna[counter:counter + 3]
        # Finds the one-letter amino acid in the dictionary
        one_letter_aa = codon_dict[codon]['1-letter']
        # Add 3 to move on to the next codon
        counter += 3
        # If it's a stop codon then end the loop
        if one_letter_aa == 'X':
            break
        # Add the amino acid to protein string
        protein += one_letter_aa
    return protein
```

```python
protein = convert_mrna_to_protein(mrna, codon_dict)
```
